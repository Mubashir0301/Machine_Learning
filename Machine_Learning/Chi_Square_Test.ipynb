```python
import pandas as pd
from scipy.stats import chi2_contingency
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
# Now suppose we want to check that survived rate is based on gender 
# Ho= Survival rate is not dpendent on gender it is by chance
# H1= Survival rate is dpendent on gender it is by chance
# We can also check by contigency table
df_contigency=pd.crosstab(df['sex'],df['survived'])
df_contigency
```

```python
#You can clearly see that there is a difference between survival rate of male and female now applt chi2 test
chi2 , p , dof , expected=chi2_contingency(df_contigency)
print("Chi2 Statistic: ",chi2)
print("P value: ",p)
print("Degree of freedom: ",dof)
print("Expected value: ",expected)
```

```python
# here is p value is very small as compared to confidence interval 0.05 so reject null hypothesis
```
