```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
#to display all columns 
pd.pandas.set_option('display.max_columns',None)
```

```python
dataset=pd.read_csv('D:/train.csv')
print(dataset.shape)
```

```python
dataset.head()
```

# Detecting Missing Values

```python
#Now we will check the percentage of missing values in each feature(column)
#First making the list of those features names having null values
features_with_null=[features for features in dataset.columns if dataset[features].isnull().sum()>1]

#Printing feature names and percentage of missing values
for feature in features_with_null:
    print(feature,np.round(dataset[feature].isnull().mean(),4),'% missing values')
    
##dataset[feature].isnull(): This correctly returns a boolean Series indicating missing values.
##.mean(): This is a method that calculates the mean of the boolean Series. 
##When you call it (with ()), it calculates the proportion of True values (which represents the proportion of missing values) as a float.
```

# Since they are many missing values, we need to find the relationship between missing values and Sales Price ¶

# Let's plot some diagram for this relationship

```python
for feature in features_with_null:
    data=dataset.copy()
    # using data we are making each column in 0,1 form where there is null value replacing that by 1 otherwise 0
    data[feature]=np.where(data[feature].isnull(),1,0) # this is like if else

    data.groupby(feature)['SalePrice'].median().plot.bar()
    plt.title(feature)
    plt.show()
```

# Numerical Variable

```python
numerical_features=[feature for feature in dataset.columns if dataset[feature].dtypes!='O'] # if they are not string object they will be numerical
print("Number of numerical features:",len(numerical_features))
dataset[numerical_features].head()
```

```python
# Seperating all features related to year
year_feature=[feature for feature in numerical_features if 'Yr' in feature or 'Year' in feature]
year_feature
```

```python
for feature in year_feature:
    print(feature,dataset[feature].unique())
```

```python
# Analyzing this temporal data
# Let's analyze the Temporal Datetime Variables
# We will check whether there is a relation between year the house is sold and SalePrice

# Group the dataset by 'YrSold', calculate the median 'SalePrice' for each year, and plot it
dataset.groupby('YrSold')['SalePrice'].median().plot()

plt.xlabel('Year Sold')
plt.ylabel('Median House Price')
plt.title('House Price vs YearSold')
plt.show()
```

```python
# Here we will compare the difference between ALL years feature with SalePrice
for feature in year_feature:
    if feature != 'YrSold': # Ensure we don't compare 'YrSold' with itself
        data = dataset.copy() # Create a copy to avoid modifying the original dataset

        # We will capture the difference between the current year variable and the year the house was sold
        # This creates a new feature representing the 'age' of the house/feature at the time of sale
        data[feature] = data['YrSold'] - data[feature]

        # Plot a scatter plot to visualize the relationship between the new 'age' feature and 'SalePrice'
        plt.scatter(data[feature], data['SalePrice'])
        plt.xlabel(feature) 
        plt.ylabel('SalePrice') 
        plt.title(f"SalePrice vs. Age of {feature} at Sale") 
        plt.show() 
```

```python
## Numerical variables are usually of 2 type
## 1. Continous variable and Discrete Variables

discrete_feature=[feature for feature in numerical_features if len(dataset[feature].unique())<25 and feature not in ['Id','SalePrice']]
print("Discrete Variables Count: {}".format(len(discrete_feature)))
```

```python
for feature in discrete_feature:
    data = dataset.copy() # Create a copy of the dataset for each iteration

    # Group the data by the current discrete feature, calculate the median SalePrice, and plot it as a bar chart
    data.groupby(feature)['SalePrice'].median().plot.bar()
    
    plt.xlabel(feature)
    plt.ylabel('SalePrice')
    plt.title(feature) 
    plt.show()
```

```python
continuous_feature = [feature for feature in numerical_features if feature not in discrete_feature and feature not in ['Id']]
print("Continuous feature Count: {}".format(len(continuous_feature)))
```

# Continous Variables

```python
# Let's analyze the continuous values by creating histograms to understand their distribution
for feature in continuous_feature:
    data = dataset.copy() # Create a copy of the dataset for each iteration

    # Plot a histogram for the current continuous feature
    # bins=25 is a common choice to visualize distribution; you can adjust this.
    data[feature].hist(bins=25)

    # Add labels and title to the plot
    plt.xlabel(feature)
    plt.ylabel('Count') # y-axis for histogram shows the count of observations in each bin
    plt.title(feature) # Use the feature name as the title for each plot

    # Display the plot
    plt.show()
```

# Exploratory Data Analysis Part 2

```python
# We will be using logarithmic transformation
for feature in continuous_feature:
    data = dataset.copy() # Create a copy of the dataset for each iteration
    
    # Check if 0 is present in the unique values of the feature
    # If 0 is present, np.log(0) is undefined (-infinity), so we skip transformation
    if 0 in data[feature].unique(): #What it does: If the condition 0 in data[feature].unique() is True (meaning the feature contains zero values),
                                    #the pass statement tells Python to do nothing and simply move to the next iteration of the loop.
                                    #Purpose: To gracefully skip the logarithmic transformation for features that would cause an error due to the presence of zeros.
        pass # Do nothing, skip transformation for this feature
    else:
        # Apply logarithmic transformation to the current continuous feature
        data[feature] = np.log(data[feature])
        
        # Apply logarithmic transformation to the SalePrice (target variable)
        # This is a common practice when the target is skewed, even if it affects the base dataset
        data['SalePrice'] = np.log(data['SalePrice'])
        
        # Plot a scatter plot to visualize the relationship after transformation
        plt.scatter(data[feature], data['SalePrice'])
        plt.xlabel(feature)
        plt.ylabel('SalePrice')
        plt.title(feature)
        plt.show()
```

### Outliers

```python
# Note to find outliers we are first converting values to their log tranformation because as data is very right skewed and 
# difficult to visualize outliers and log transformation compress the larger values so it will help us to see outliers

for feature in continuous_feature:
    data = dataset.copy() 
    # Check if 0 is present in the unique values of the feature
    # If 0 is present, np.log(0) is undefined, so we skip transformation
    if 0 in data[feature].unique():
        pass # Do nothing, skip transformation for this feature
    else:
        # Apply logarithmic transformation to the current continuous feature
        data[feature] = np.log(data[feature]) 
        
        # Create a box plot for the transformed feature
        # It's good practice to explicitly use plt.boxplot if you're mixing with matplotlib calls,
        # but the Pandas method is also valid.
        data.boxplot(column=feature) 
        
        # Add labels and title to the plot
        plt.ylabel(feature) 
        plt.title(feature) 
        
        # Display the plot
        plt.show()
```
