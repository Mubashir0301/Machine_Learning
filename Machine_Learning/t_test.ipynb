# We know that t_test checks the variance of data from the mean

```python
import seaborn as sns
from scipy.stats import ttest_1samp
import pandas as pd
```

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
df1=df['age']
df1.describe()
```

```python
# lets check the age and compare with a known value
ttest_1samp(df1,50)
#After running you will see that it will not give statistics and p value bcz if you observe the data it is dispersed and not normal
```

```python
stats , p=ttest_1samp(df1,50)
print("stats=%.3f , p=%.3f "% (stats,p))
#Checking with confidence interval
if (p>0.05):
    print("Distribution is probably the same") # means 50(data point) and mean of df1 has no difference
else (p>0.05):
    print("Distribution is not same")
#probably you might not get values bcz data is not normalized but if you normalize it you will get values
```
