```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load a clean dataset
iris = load_iris()
df = pd.DataFrame(data=np.c_[iris['data'], iris['target']],
                  columns=iris['feature_names'] + ['target'])

df.loc[np.random.choice(df.index, 10), 'sepal length (cm)'] = np.nan
df.loc[np.random.choice(df.index, 10), 'petal width (cm)'] = np.nan

print("Dataset with introduced missing values:\n", df.isnull().sum())
```

```python
sns.heatmap(df.isnull())
plt.show()
```

```python
sns.boxplot(x="sepal length (cm)",data=df)
```

```python
# No outliers
```

```python
sns.boxplot(x="petal width (cm)",data=df)
```

```python
#Replacing missing values with medians. We are save to go with mean also as there are no outliers
```

```python
median_sepal_value=df["sepal length (cm)"].median()
```

```python
print(median_sepal_value)
```

```python
df["sepal length (cm)"].fillna(median_sepal_value,inplace =True)
```

```python
sns.heatmap(df.isnull())
plt.show()
```

```python
median_petal_value=df["petal width (cm)"].median()
df["petal width (cm)"].fillna(median_petal_value,inplace =True)
```

```python
sns.heatmap(df.isnull())
plt.show()
```
