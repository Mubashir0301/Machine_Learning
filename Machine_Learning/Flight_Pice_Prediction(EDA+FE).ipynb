```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
train_df = pd.read_excel(r"D:\ML_Files\Data_Train.xlsx")
train_df.head()
```

```python
test_df = pd.read_excel(r"D:\ML_Files\Test_set.xlsx")
test_df.head()
```

```python
df = pd.concat([train_df, test_df], ignore_index=True)
df.head()
```

```python
## Fetaure Engineering Process
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
## OR
df["Date"] = df['Date_of_Journey'].apply(lambda x:x.split("/")[0])
df["Month"] = df['Date_of_Journey'].apply(lambda x:x.split("/")[1])
df["Year"] = df['Date_of_Journey'].apply(lambda x:x.split("/")[2])
```

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
```

```python
df.drop('Date_of_Journey',axis=1 , inplace=True)
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
```

```python
df['Arrival_Hour']=df['Arrival_Time'].apply(lambda x: x.split(':')[0])
df['Arrival_Minutes']=df['Arrival_Time'].apply(lambda x: x.split(':')[1])
```

```python
df['Arrival_Hour'] = df['Arrival_Hour'].astype(int)
df['Arrival_Minutes'] = df['Arrival_Minutes'].astype(int)
```

```python
df.info()
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
# Same for departure time


# The .str accessor is excellent for vectorized string operations
# It safely handles nulls without raising an error
df['Dep_Time_Hour'] = df['Dep_Time'].str.split(':').str[0]
df['Dep_Time_Minutes'] = df['Dep_Time'].str.split(':').str[1]

# Now, convert to numeric type. pd.to_numeric() with errors='coerce' is a robust way.
# It will convert non-numeric strings (e.g., '15 ', '10.0') to NaN.
df['Dep_Time_Hour'] = pd.to_numeric(df['Dep_Time_Hour'], errors='coerce')
df['Dep_Time_Minutes'] = pd.to_numeric(df['Dep_Time_Minutes'], errors='coerce')

# Now that the data is numeric, you can safely fill NaNs (if any) and convert to integer.
# It's common to fill NaNs with 0, but this depends on your data context.
df['Dep_Time_Hour'] = df['Dep_Time_Hour'].fillna(0).astype(int)
df['Dep_Time_Minutes'] = df['Dep_Time_Minutes'].fillna(0).astype(int)

# Finally, drop the original column
df.drop('Dep_Time', axis=1, inplace=True)
```

```python
df['Additional_Info'].unique()
```

```python
df['Duration_Hour']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```

```python
df[df['Duration_Hour']=='5m']
```

```python
stops_map = {
    'non-stop': 0,
    '1 stop': 1,
    '2 stops': 2,
    '3 stops': 3,
    '4 stops': 4,

    np.nan: 2 
}

# 2. Apply the map to the column
df['Total_Stops'] = df['Total_Stops'].map(stops_map)

# 3. Now that the column contains only numbers (and possibly NaNs if a label wasn't mapped),
#    fill any remaining NaNs with a default value, then convert to int.
df['Total_Stops'] = df['Total_Stops'].fillna(2).astype(int)
print(df['Total_Stops'].isnull().sum())
```

```python
#Dropping this row bcz due to this we are not able to convert Duration_Hour to int
# Find the index of the row where 'Duration' is '5m'
index_to_drop = df[df['Duration'] == '5m'].index

# Drop the row using its new index
df.drop(index_to_drop, axis=0, inplace=True)
```

```python
df['Duration_Hour'] = df['Duration_Hour'].astype(int)
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
labelencoder = LabelEncoder()

df['Airline'] = labelencoder.fit_transform(df['Airline'])
df['Source'] = labelencoder.fit_transform(df['Source'])
df['Destination'] = labelencoder.fit_transform(df['Destination'])
df['Additional_Info'] = labelencoder.fit_transform(df['Additional_Info'])
```

```python
#I used prefix='Destination' when demonstrating one-hot encoding with pd.get_dummies() for a specific reason.
#The prefix parameter adds a string prefix to the new binary columns that are created. For example, if your original column is named 'Destination' and its unique categories are 'Banglore', 'New Delhi', and 'Cochin', using prefix='Destination' will result in new columns named:
#Destination_Banglore
#Destination_New Delhi
#Destination_Cochin
```

```python
destination_dummies = pd.get_dummies(df['Destination'], prefix='Destination', drop_first=True)
df = pd.concat([df, destination_dummies], axis=1)
```

```python
df.drop('Destination', axis=1, inplace=True)
```

```python
df.head(20)
```

```python
boolean_to_int_map = {True: 1, False: 0}

# Select all the one-hot encoded columns
new_destination_columns = [col for col in df.columns if col.startswith('Destination_')]
df[new_destination_columns] = df[new_destination_columns].apply(
    lambda col: col.map(boolean_to_int_map)
)
```
