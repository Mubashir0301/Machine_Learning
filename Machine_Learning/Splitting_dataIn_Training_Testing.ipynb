```python
X=df[['Mileage','Age(yrs)']]
Y=df['Sell Price($)']
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,Y_test=train_test_split(X,Y,test_size=0.2) 
#Now 20% data will be for testing and 80% for training if you print len of X_train you will always get 10 random rows because it is training data which is 80%
len(X_train)
```
