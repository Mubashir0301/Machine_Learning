```python
import pandas as pd
import numpy as np

# --- 1. Setup: Create Practice DataFrames ---

# DataFrame A (like 'ratings'): Contains core IDs and a value
data_A = pd.DataFrame({
    'Transaction-ID': [101, 102, 103, 104, 105, 106],
    'User-ID': ['U1', 'U1', 'U2', 'U3', 'U4', 'U5'],
    'Product-ID': ['P1', 'P2', 'P1', 'P3', 'P5', 'P6'], # P5 and P6 don't exist in C
    'Rating': [5, 4, 5, 3, 1, 4]
})

# DataFrame B (like 'users'): Contains User details
data_B = pd.DataFrame({
    'User-ID': ['U1', 'U2', 'U3', 'U5'], # U4 is missing here
    'Location': ['NY', 'LA', 'MIA', 'CHI'],
    'Age': [35, 22, 41, np.nan] # Note the missing Age
})

# DataFrame C (like 'books'): Contains Product details
data_C = pd.DataFrame({
    'Product-ID': ['P1', 'P2', 'P3', 'P4'],
    'Product-Name': ['Laptop', 'Mouse', 'Monitor', 'Keyboard'],
    'Price': [1200, 25, 300, 75]
})
```

```python
merged_data = data_A.merge(data_B, on='User-ID', how='left')
merged_data = merged_data.merge(data_C, on='Product-ID', how='left')
```
