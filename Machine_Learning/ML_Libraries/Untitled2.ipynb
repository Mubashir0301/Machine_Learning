```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("D:/canada_per_capita_income.csv")

# Rename the column to remove spaces and special characters
df.rename(columns={"per capita income (US$)": "income"}, inplace=True)

# Plot the data
plt.figure(figsize=(10,5))
plt.plot(df["year"], df["income"], marker='o', linestyle='-', color='b', label="Per Capita Income")

# Labels and Title
plt.xlabel("Year")
plt.ylabel("Per Capita Income (US$)")
plt.title("Canada's Per Capita Income Over the Years")
plt.legend()
plt.grid(True)

# Show the graph
plt.show()
```
