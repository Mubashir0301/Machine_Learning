```python
df['children']=df['children'].fillna(0)
```

```python
df['children'].value_counts()
```

```python
df.drop('company',axis=1,inplace=True)
```

```python

most_frequent = df.groupby('country')['market_segment'].agg(lambda x: x.mode()[0] if not x.mode().empty else "Unknown")
print(most_frequent)
```

```python


# Fill the NaNs based on the country's specific mode
df['market_segment'] = df['market_segment'].fillna(most_frequent)
df['market_segment']
```

```python
df.isnull().sum()
```

```python
#now I am filling the nulls of country with the mode based on the market segment
country_modes = df.groupby('market_segment')['country'].transform(lambda x: x.mode()[0] if not x.mode().empty else "Unknown")

df['country'] = df['country'].fillna(country_modes)
```

```python
df['country'].isnull().sum()
```

```python
# Now null value of agent means that there is no agent involved in the booking so we can fill those nulls with 0
df['agent']=df['agent'].fillna(0)
```

```python
df['agent'].isnull().sum()
```

```python
# Checking that which country prefers which type of meal

top_10_countries = df['country'].value_counts().head(10).index

# Filter the dataframe to only include these countries
df_top_10 = df[df['country'].isin(top_10_countries)]
meal_counts = df_top_10.groupby(['country', 'meal'])['meal'].size()
print(meal_counts)
```

### Q1: What is the overall percentage of bookings that get cancelled? Furthermore, which type of hotel (City Hotel or Resort Hotel) has a higher cancellation rate?

```python
avg_percentage=df['is_canceled'].mean()*100
print(f"Average percentage of canceled bookings: {avg_percentage:.2f}%")

df.groupby('hotel')['is_canceled'].mean()*100
```

```python
hotels=df['hotel'].value_counts().reset_index()
print(hotels)
```

```python

hotel_cancellation = df.groupby('hotel')['is_canceled'].mean() * 100
hotel_cancellation = hotel_cancellation.reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='hotel', y='is_canceled', data=hotel_cancellation)

plt.title('Cancellation Rate by Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.show()
```

### Its mean city hotel has high cancellation rate compared to Resort

### Q2: Is there a relationship between how far in advance a guest books (lead_time) and whether they cancel (is_canceled)?

### Q3: Which Market Segment (e.g., Online TA, Groups, Direct) generates the highest ADR (Average Daily Rate)?

```python
highest_ADR=df.groupby('market_segment')['adr'].mean().sort_values(ascending=False)
print(highest_ADR)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='market_segment', y='adr', data=highest_ADR.reset_index())
plt.title('Average ADR by Market Segment')
plt.show()
```

### Q4: Which month has the highest number of guests arriving? Is the "peak season" different for the City Hotel versus the Resort Hotel?

```python
df['arrival_date_month'].value_counts().reset_index()
```

```python
df.groupby('hotel')['arrival_date_month'].value_counts()

# In both hotels the month of August has the highest number of bookings
```

```python
# 1. Define the correct order of months
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']

# 2. Convert the column to a 'Category' type with that specific order
df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories=month_order, ordered=True)

# 3. Now, plotting is much more professional
plt.figure(figsize=(12,6))
sns.countplot(x='arrival_date_month', hue='hotel', data=df)
plt.title('Monthly Arrival Comparison')
plt.xticks(rotation=45) # Tilt the labels so they don't overlap
plt.show()
```

### Which means though in august most guest arrives but most of them choose city hotel

### Q5: What is the most common length of stay (total nights)? Does this differ between the City Hotel (short business trips) and the Resort Hotel (longer vacations)?

```python
# 1. Create the column
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

# 2. Filter for stays between 1 and 10 nights for a clean view
stay_data = df[df['total_nights'].between(1, 10)]

# 3. Plot it
plt.figure(figsize=(15,8))
sns.countplot(x='total_nights', hue='hotel', data=stay_data)

plt.title('Distribution of Stay Length (1-10 Nights)')
plt.xlabel('Number of Nights')
plt.ylabel('Number of Bookings')
plt.show()

```
